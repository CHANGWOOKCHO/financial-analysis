```python
import os
import requests
import bs4
import pandas as pd
```

#### 크롤링 request header Info

```python
my_headers = {
    "user-agent" : "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/140.0.0.0 Safari/537.36"
}
```

#### S&P500 기업리스트 csv저장

```python
#현재 디렉토리 경로
vspwd = os.getcwd()
# 디렉토리 재설정
os.chdir(vspwd)
```

```python
url = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"
res = requests.get(url, headers=my_headers)

soup = bs4.BeautifulSoup(res.text)

tables = soup.find_all('table')
snapshot_table = soup.find('table', attrs={'id': 'constituents'})
tables = pd.read_html(str(snapshot_table))
df = tables[0]

#기업정보목록 확인
df.head()

df.to_csv("companyList.csv", encoding='utf-8', index=true)
```
